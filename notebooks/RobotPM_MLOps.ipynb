## 1. Original Architecture

The project initially followed a notebook-centered machine learning workflow where most tasks were performed directly inside Jupyter notebooks.

The original workflow included:

- Loading datasets directly inside notebook cells
- Cleaning and exploring data inside the notebook
- Defining the train/test split directly in the notebook
- Scaling features inside the notebook
- Implementing gradient descent directly in notebook cells
- Calculating evaluation metrics directly in the notebook
- Hard-coding experiment parameters such as learning rate, iterations, feature selection, and test size

```text
Data Sources
     |
     v
Jupyter Notebook
     |
     +-- Data Loading
     +-- Preprocessing
     +-- Model Training
     +-- Evaluation
     +-- Visualization
     |
     v
Model Results

## 2. Architecture Changes Implemented

The project was refactored from a notebook-centered workflow into a modular machine learning architecture.

### Modular Components

Reusable functionality was moved into the `src/` directory:

- `data_loader.py` — handles loading data from CSV files, APIs, and a PostgreSQL database.
- `preprocessing.py` — handles train/test splitting and feature scaling.
- `model.py` — contains the univariate linear regression implementation from scratch using gradient descent.
- `evaluation.py` — calculates MAE, RMSE, and R².

### Configuration Management

Experiment settings were moved to:

`configs/experiment_config.yaml`

The configuration file controls values such as:

- Dataset paths
- Selected feature and target
- Test size
- Random state
- Learning rate
- Number of iterations

This reduces hard-coded experiment settings inside the notebook.

### Experiment Tracking

Experiment results are stored in:

`experiments/results.csv`

Each experiment records the model configuration and evaluation metrics, along with an experiment ID and timestamp.

### Database and API Integration

The data-loading architecture supports multiple data sources:

- Local CSV files
- Statistics Canada API
- Neon-hosted PostgreSQL database

Database credentials are stored using environment variables rather than being hard-coded into the source code.

## 3. Updated MLOps Architecture

The updated architecture separates data ingestion, preprocessing, model training, evaluation, configuration, and experiment tracking into dedicated components.

### Updated Workflow

```text
                     experiment_config.yaml
                              |
                              v
Data Sources ---> data_loader.py
   |                  |
   |                  v
   |           preprocessing.py
   |                  |
   |                  v
   |              model.py
   |                  |
   |                  v
   |            evaluation.py
   |                  |
   |                  v
   |          experiments/results.csv
   |
   +-- CSV
   +-- Statistics Canada API
   +-- Neon PostgreSQL

## 4. Benefits of the Updated Architecture

The updated architecture improves the project in several ways.

### Modularity

Each component has a specific responsibility. Data loading, preprocessing, modeling, and evaluation can be modified independently without rewriting the entire notebook.

### Reusability

Functions and classes stored in the `src/` directory can be imported into different notebooks or future applications instead of duplicating code.

### Configurability

Experiment parameters are stored in `experiment_config.yaml`. This allows different experiments to be performed by changing configuration values rather than modifying the model code.

### Reproducibility

A fixed random state, configuration file, and pinned dependencies in `requirements.txt` make it easier to reproduce an experiment using the same settings.

### Experiment Tracking

Experiment IDs, timestamps, model parameters, and evaluation metrics are stored in `experiments/results.csv`, allowing multiple experiment runs to be reviewed and compared.

### Security

Database credentials are stored in environment variables through a `.env` file rather than being hard-coded into notebooks or Python modules.

### Maintainability

Separating reusable logic from exploratory notebooks makes the project easier to understand, debug, test, and extend as additional machine learning functionality is introduced.

## 5. Recommended Additions and Future Enhancements

The current architecture provides a foundation for a reproducible machine learning workflow. As the project grows, the following additions could improve the architecture further.

### Automated Testing

A `tests/` directory could be introduced to test individual modules such as data loading, preprocessing, model prediction, and evaluation.

Example future structure:

```text
tests/
├── test_data_loader.py
├── test_preprocessing.py
├── test_model.py
└── test_evaluation.py

## 6. Example Enhancement Breakdown

One possible future enhancement is automated testing. The current modular architecture makes testing easier because each responsibility is already separated into individual Python files.

### Example: Testing the Preprocessing Module

A future test file could be created as:

`tests/test_preprocessing.py`

Example:

```python
from src.preprocessing import split_data

def test_split_data():
    X_train, X_test, y_train, y_test = split_data(
        X,
        y,
        test_size=0.20,
        random_state=42
    )

    assert len(X_train) == len(y_train)
    assert len(X_test) == len(y_test)
```
